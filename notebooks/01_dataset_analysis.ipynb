```python
from pathlib import Path
import pandas as pd

# Locate project root and raw dataset folder
current = Path.cwd()

possible_paths = [
    current / "datasets" / "raw",
    current.parent / "datasets" / "raw",
    current / "data" / "raw",
    current.parent / "data" / "raw",
]

RAW_DIR = next((p for p in possible_paths if p.exists()), None)

if RAW_DIR is None:
    raise FileNotFoundError(
        "Could not find datasets/raw or data/raw. "
        "Open the notebook from inside the Jalachakra project."
    )

print("RAW DATA FOLDER:")
print(RAW_DIR)
print("\nFILES FOUND:\n")

files = sorted([f for f in RAW_DIR.iterdir() if f.is_file()])

for f in files:
    print(f.name)

print(f"\nTotal files: {len(files)}")
```

```python
# STEP 1B — Load all raw datasets safely

datasets = {}

for file in files:
    try:
        extension = file.suffix.lower()

        if extension == ".csv":
            try:
                df = pd.read_csv(file, encoding="utf-8")
            except UnicodeDecodeError:
                print(f"UTF-8 failed for {file.name} → trying latin1")
                df = pd.read_csv(file, encoding="latin1")

        elif extension in [".xlsx", ".xls"]:
            df = pd.read_excel(file)

        else:
            continue

        datasets[file.name] = df

    except Exception as e:
        print(f"ERROR reading {file.name}: {e}")

print("\nSuccessfully loaded datasets:")
print("=" * 70)

for name, df in datasets.items():
    print(f"{name:45} {df.shape}")

print("\nTotal successfully loaded:", len(datasets))
```

```python
# STEP 1C — Inspect columns and data types

for name, df in datasets.items():

    print("\n" + "=" * 80)
    print(f"DATASET: {name}")
    print("=" * 80)

    print(f"Rows    : {df.shape[0]}")
    print(f"Columns : {df.shape[1]}")

    print("\nColumns and data types:")

    column_info = pd.DataFrame({
        "column": df.columns,
        "dtype": df.dtypes.astype(str).values,
        "missing": df.isna().sum().values,
        "unique_values": df.nunique(dropna=True).values
    })

    display(column_info)
```

```python
# STEP 1D — Missing values and duplicate analysis

missing_summary = []
duplicate_summary = []

for name, df in datasets.items():

    # -----------------------------
    # Missing-value information
    # -----------------------------
    total_missing = int(df.isna().sum().sum())

    missing_columns = int(
        (df.isna().sum() > 0).sum()
    )

    missing_summary.append({
        "dataset": name,
        "rows": len(df),
        "total_missing_cells": total_missing,
        "columns_with_missing_values": missing_columns
    })

    # -----------------------------
    # Duplicate information
    # -----------------------------
    duplicate_rows = int(df.duplicated().sum())

    duplicate_summary.append({
        "dataset": name,
        "rows": len(df),
        "duplicate_rows": duplicate_rows
    })


missing_df = pd.DataFrame(missing_summary)
duplicate_df = pd.DataFrame(duplicate_summary)

print("MISSING VALUE SUMMARY")
display(missing_df)

print("\nDUPLICATE SUMMARY")
display(duplicate_df)
```

```python
# STEP 1D-B — Exact columns containing missing values

for name, df in datasets.items():

    missing = df.isna().sum()
    missing = missing[missing > 0]

    if len(missing) > 0:

        print("\n" + "=" * 70)
        print(name)
        print("=" * 70)

        for column, count in missing.items():

            percentage = (count / len(df)) * 100

            print(
                f"{column}: "
                f"{count} missing "
                f"({percentage:.2f}%)"
            )
```

```python
# STEP 1E — Date, location, crop and time-coverage analysis

DATE_KEYWORDS = [
    "date",
    "datetime",
    "time",
    "year",
    "month"
]

LOCATION_KEYWORDS = [
    "state",
    "district",
    "location",
    "region",
    "latitude",
    "longitude",
    "lat",
    "lon"
]

CROP_KEYWORDS = [
    "crop",
    "crops",
    "crop_type",
    "label"
]

for name, df in datasets.items():

    print("\n" + "=" * 90)
    print(f"DATASET: {name}")
    print("=" * 90)

    # ---------------------------
    # Date-like columns
    # ---------------------------
    date_columns = [
        col for col in df.columns
        if any(
            keyword in str(col).lower()
            for keyword in DATE_KEYWORDS
        )
    ]

    if date_columns:
        print("\nDATE/TIME COLUMNS:")

        for col in date_columns:

            print(f"\nColumn: {col}")

            series = df[col].dropna()

            print("Data type:", df[col].dtype)
            print("Unique values:", series.nunique())

            # Try to interpret as dates
            converted = pd.to_datetime(
                series,
                errors="coerce"
            )

            valid_dates = converted.dropna()

            if len(valid_dates) > 0:
                print("Possible minimum date:", valid_dates.min())
                print("Possible maximum date:", valid_dates.max())
            else:
                print("Could not interpret this column as dates.")

    else:
        print("\nNo obvious date/time columns found.")

    # ---------------------------
    # Location columns
    # ---------------------------
    location_columns = [
        col for col in df.columns
        if any(
            keyword in str(col).lower()
            for keyword in LOCATION_KEYWORDS
        )
    ]

    if location_columns:
        print("\nLOCATION COLUMNS:")

        for col in location_columns:

            print(f"\nColumn: {col}")
            print("Unique values:", df[col].nunique(dropna=True))

            print(
                df[col]
                .dropna()
                .astype(str)
                .value_counts()
                .head(15)
            )

    else:
        print("\nNo obvious location columns found.")

    # ---------------------------
    # Crop columns
    # ---------------------------
    crop_columns = [
        col for col in df.columns
        if any(
            keyword in str(col).lower()
            for keyword in CROP_KEYWORDS
        )
    ]

    if crop_columns:
        print("\nCROP COLUMNS:")

        for col in crop_columns:

            print(f"\nColumn: {col}")
            print(
                "Unique crop values:",
                df[col].nunique(dropna=True)
            )

            print(
                df[col]
                .dropna()
                .astype(str)
                .value_counts()
                .head(20)
            )

    else:
        print("\nNo obvious crop columns found.")
```

```python
# STEP 1E — Compact dataset coverage summary
# This analyzes ALL datasets without producing huge output.

summary_rows = []

for name, df in datasets.items():

    # ---------- Date detection ----------
    date_columns = []

    for col in df.columns:
        col_lower = str(col).lower()

        if any(word in col_lower for word in [
            "date", "datetime", "time"
        ]):
            date_columns.append(col)

    date_start = None
    date_end = None

    for col in date_columns:

        converted = pd.to_datetime(
            df[col],
            errors="coerce"
        )

        valid = converted.dropna()

        # Ignore columns that are not really date-like
        if len(valid) >= max(3, int(len(df) * 0.5)):

            current_start = valid.min()
            current_end = valid.max()

            if date_start is None or current_start < date_start:
                date_start = current_start

            if date_end is None or current_end > date_end:
                date_end = current_end

    # ---------- Crop detection ----------
    crop_columns = []

    for col in df.columns:
        col_lower = str(col).lower()

        if any(word in col_lower for word in [
            "crop", "crops", "crop_type", "label"
        ]):
            crop_columns.append(col)

    crop_count = None

    if crop_columns:
        counts = []

        for col in crop_columns:
            counts.append(df[col].nunique(dropna=True))

        crop_count = max(counts)

    # ---------- Location detection ----------
    location_columns = []

    for col in df.columns:
        col_lower = str(col).lower()

        if any(word in col_lower for word in [
            "state",
            "district",
            "location",
            "region",
            "latitude",
            "longitude",
            "lat",
            "lon"
        ]):
            location_columns.append(col)

    location_count = None

    if location_columns:
        counts = []

        for col in location_columns:
            counts.append(df[col].nunique(dropna=True))

        location_count = max(counts)

    # ---------- Important CWR variables ----------
    cwr_keywords = [
        "water",
        "requirement",
        "evapotranspiration",
        "et0",
        "etc",
        "kc",
        "rainfall",
        "precip",
        "temperature",
        "humidity",
        "wind",
        "solar",
        "moisture",
        "irrigation",
        "growth",
        "stage",
        "root",
        "whc"
    ]

    cwr_columns = []

    for col in df.columns:

        normalized = str(col).lower().replace("_", " ")

        if any(word in normalized for word in cwr_keywords):
            cwr_columns.append(str(col))

    summary_rows.append({
        "dataset": name,
        "rows": len(df),
        "columns": len(df.columns),
        "date_columns": ", ".join(map(str, date_columns))
            if date_columns else "None",
        "date_start": date_start,
        "date_end": date_end,
        "crop_columns": ", ".join(map(str, crop_columns))
            if crop_columns else "None",
        "crop_count": crop_count,
        "location_columns": ", ".join(map(str, location_columns))
            if location_columns else "None",
        "location_count": location_count,
        "cwr_related_columns": ", ".join(cwr_columns)
            if cwr_columns else "None"
    })


coverage_df = pd.DataFrame(summary_rows)

coverage_df
```

```python
# STEP 1E-C — Correct date and crop interpretation

corrected_rows = []

for name, df in datasets.items():

    # -----------------------------
    # REAL DATE COLUMNS
    # -----------------------------
    real_date_columns = []

    for col in df.columns:
        col_lower = str(col).lower()

        # Only treat these as true date fields
        if col_lower in ["date", "datetime", "timestamp"]:
            real_date_columns.append(col)

    # Ignore integer DATE/MONTH helper fields unless they
    # can actually be interpreted as calendar dates.
    date_start = pd.NaT
    date_end = pd.NaT
    selected_date_column = "None"

    for col in real_date_columns:

        converted = pd.to_datetime(
            df[col],
            errors="coerce"
        )

        valid = converted.dropna()

        if len(valid) >= max(3, int(len(df) * 0.5)):

            selected_date_column = col
            date_start = valid.min()
            date_end = valid.max()
            break

    # -----------------------------
    # REAL CROP COLUMNS
    # -----------------------------
    real_crop_columns = []

    for col in df.columns:

        col_lower = str(col).lower().strip()

        if (
            col_lower in ["crop", "crop_type", "crops", "label"]
            or col_lower.startswith("crop ")
            or col_lower.endswith(" crop")
        ):
            real_crop_columns.append(col)

    crop_count = None

    if real_crop_columns:

        crop_counts = [
            df[col].nunique(dropna=True)
            for col in real_crop_columns
        ]

        crop_count = max(crop_counts)

    corrected_rows.append({
        "dataset": name,
        "rows": len(df),
        "real_date_column": selected_date_column,
        "date_start": date_start,
        "date_end": date_end,
        "crop_columns": ", ".join(
            map(str, real_crop_columns)
        ) if real_crop_columns else "None",
        "crop_count": crop_count
    })


corrected_coverage_df = pd.DataFrame(corrected_rows)

display(corrected_coverage_df)
```

```python
# STEP 1F — Preliminary scientific role classification

role_rows = []

for name, df in datasets.items():

    columns = [
        str(c).lower().strip()
        for c in df.columns
    ]

    text = " ".join(columns)

    has_weather = any(
        x in text
        for x in [
            "temperature", "temp", "humidity",
            "wind", "solar", "radiation",
            "t2m", "rh2m", "ws2m"
        ]
    )

    has_rainfall = any(
        x in text
        for x in ["rainfall", "precipitation", "precip"]
    )

    has_soil = any(
        x in text
        for x in [
            "soil", "nitrogen", "phosphorous",
            "phosphorus", "potassium", "ph", "moisture"
        ]
    )

    has_crop = any(
        x in columns
        for x in ["crop", "crop_type", "crops", "label"]
    )

    has_growth_stage = any(
        x in text
        for x in [
            "growth_stage", "growth stage",
            "stage", "duration"
        ]
    )

    has_water_target = any(
        x in text
        for x in [
            "water_requirement",
            "water requirement",
            "water_required",
            "water required",
            "water_requirement_liters",
            "evapotranspiration",
            "et0", "etc", "kc"
        ]
    )

    has_irrigation = "irrigation" in text

    has_market = any(
        x in text
        for x in ["price", "market", "commodity", "arrival"]
    )

    if (
        "karnataka_crop_growth_stage_reference" in name.lower()
        or "crop_calender" in name.lower()
    ):
        role = "AGRONOMIC_REFERENCE"

    elif has_market:
        role = "MARKET"

    elif "crop_recommendation" in name.lower():
        role = "CROP_RECOMMENDATION"

    elif has_weather and has_rainfall:
        role = "WEATHER_RAINFALL"

    elif has_soil and not has_weather:
        role = "SOIL_SUPPORTING"

    elif has_water_target and has_crop and has_growth_stage:
        role = "CWR_CANDIDATE"

    elif "data_season" in name.lower():
        role = "SUPPORTING_REVIEW_REQUIRED"

    elif "crop_dataset" in name.lower():
        role = "SUPPORTING_REVIEW_REQUIRED"

    else:
        role = "REVIEW_REQUIRED"

    role_rows.append({
        "dataset": name,
        "preliminary_role": role,
        "has_crop": has_crop,
        "has_soil": has_soil,
        "has_weather": has_weather,
        "has_rainfall": has_rainfall,
        "has_growth_stage": has_growth_stage,
        "has_water_target": has_water_target,
        "has_irrigation": has_irrigation,
        "has_market": has_market
    })

role_df = pd.DataFrame(role_rows)

display(role_df)
```

```python
# STEP 2C — Physical consistency check for water_requirement_liters

df = datasets["crop_water_data.csv"].copy()

# Expected water volume:
# 1 mm of water over 1 hectare = 10,000 litres
df["expected_volume_from_ET_mm"] = (
    df["evapotranspiration"]
    * df["field_area_ha"]
    * 10000
)

# Ratio between supplied target and ET-based volume
df["target_to_physics_ratio"] = (
    df["water_requirement_liters"]
    / df["expected_volume_from_ET_mm"].replace(0, np.nan)
)

print("PHYSICAL CONSISTENCY CHECK")
print("=" * 70)

print("\nTarget statistics:")
display(
    df["water_requirement_liters"].describe()
)

print("\nEvapotranspiration statistics:")
display(
    df["evapotranspiration"].describe()
)

print("\nField area statistics:")
display(
    df["field_area_ha"].describe()
)

print("\nTarget / physics ratio:")
display(
    df["target_to_physics_ratio"].describe()
)

print("\nCorrelation with target:")

correlation_columns = [
    "evapotranspiration",
    "field_area_ha",
    "crop_age_days",
    "soil_moisture_pct",
    "temperature",
    "humidity",
    "rainfall_mm",
    "water_requirement_liters"
]

display(
    df[correlation_columns].corr()["water_requirement_liters"]
    .sort_values(ascending=False)
)
```

```python
# STEP 5H — FIND EXACT NASA POWER / WEATHER DOWNLOAD CODE

from pathlib import Path
import json

PROJECT_ROOT = Path(
    r"C:\Users\bhavy\OneDrive\Pictures\Desktop\AgriHydro"
)

NOTEBOOK = (
    PROJECT_ROOT
    / "notebooks"
    / "01_dataset_analysis.ipynb"
)

with open(NOTEBOOK, "r", encoding="utf-8") as f:
    notebook = json.load(f)

print("=" * 100)
print("JALACHAKRA — EXACT WEATHER API / DOWNLOAD CODE")
print("=" * 100)

strong_keywords = [
    "power.larc.nasa.gov",
    "https://power",
    "requests.get",
    "requests.request",
    "NASA POWER",
    "POWER API",
    "api/temporal",
    "parameters=",
    "latitude=",
    "longitude="
]

found_cells = []

for cell_number, cell in enumerate(
    notebook.get("cells", []),
    start=1
):

    source = "".join(cell.get("source", []))

    if not source:
        continue

    matches = []

    for keyword in strong_keywords:
        if keyword.lower() in source.lower():
            matches.append(keyword)

    if matches:
        found_cells.append(
            (cell_number, cell.get("cell_type"), matches, source)
        )

print("\nMatching cells:", len(found_cells))

for cell_number, cell_type, matches, source in found_cells:

    print("\n" + "=" * 100)
    print(f"CELL {cell_number} — {cell_type}")
    print("Matched keywords:", matches)
    print("-" * 100)

    # Print the complete cell so URL/coordinates are visible
    print(source)

print("\n" + "=" * 100)
print("WHAT WE NEED")
print("=" * 100)

print("""
Look for:
1. power.larc.nasa.gov
2. latitude
3. longitude
4. start/end dates
5. parameters such as T2M, RH2M, WS2M, PRECTOTCORR
6. code that saves the response as Climate_Data.csv
""")

print("\nDo NOT calculate ET0 yet.")
```
